# Bundesliga archive: coverage and feature engineering
Audit complete season schedules, then inspect lagged pre-match covariates rather than post-match box scores.

```
from pathlib import Path
import json, sys, pandas as pd
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
archive = {y: json.loads((root / f'public/data/{y}.json').read_text())['matches'] for y in range(2021, 2026)}
for year, games in archive.items():
    df = pd.DataFrame(games)
    assert len(df) == 306 and df.md.nunique() == 34 and df.h.nunique() == 18
df.groupby('md')[['hx','ax','hg','ag']].mean().head()
```
